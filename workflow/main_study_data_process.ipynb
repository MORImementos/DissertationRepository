```python
import pandas as pd
```

```python
pd.read_excel("data/main_study_data.xlsx")
```

going to get the names of each, iterate through them, and use them as the names for the files

```python
xl = pd.ExcelFile("data/main_study_data.xlsx")
xl.sheet_names
```

```python
f_names = []
for i in xl.sheet_names:
    file_name = ""
    if "dat" in i.lower():
        file_name = i.replace("(", "_").replace(")", "").replace(" ", "_").replace("_data", "").replace("_dat", "").replace("_Data", "").lower()
        f_names.append(file_name)
```

```python
df_dict = {
    sheet_name: xl.parse(sheet_name).ffill()
    for sheet_name in xl.sheet_names
}
```

```python
for sheet_name, df in df_dict.items():
    if "dat" not in sheet_name.lower():
        continue
    df = df.rename(columns={"lemma": "lemma_1"})
    print(df)
    df = df[[
        "lemma_1",
        "lemma_2",
        "year",
        "genre",
        "source",
        "data"
        ]]
    file_name = (
        sheet_name
        .replace("(", "_")
        .replace(")", "")
        .replace(" ", "_")
        .replace("_data", "")
        .replace("_dat", "")
        .replace("_Data", "")
        .lower()
    )

    df.to_csv(f"data/main/{file_name}.csv", index=False)
```

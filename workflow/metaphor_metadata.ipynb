```python
# === 1) produce primary_frames.txt ===
import ast
import re
import pandas as pd

# If df isn't already loaded, load it:
# df = pd.read_csv("path/to/your.csv")

def is_primary_only(x):
    """Return True only if x represents exactly ['Primary'] or the string 'Primary'."""
    if pd.isna(x):
        return False
    if x == "Primary":
        return True
    if isinstance(x, list):
        return x == ["Primary"]
    if isinstance(x, str):
        # Try literal_eval for list-like strings: "['Primary']"
        try:
            val = ast.literal_eval(x)
            return val == ["Primary"]
        except Exception:
            # Fall back to delimiter-splitting (e.g. "Primary; Complex" -> not exclusive)
            parts = re.split(r'[;,]\s*', x.strip())
            parts = [p for p in parts if p]
            return len(parts) == 1 and parts[0] == "Primary"
    return False

# mask rows that are exclusively Primary
mask = df["hasMetaphorType"].apply(is_primary_only)

# collect frames that appear in source or target columns within those rows
src = df.loc[mask, "hasSourceFrame"].dropna()
tgt = df.loc[mask, "hasTargetFrame"].dropna()


def normalize_frame_name(x):
    # Case 1: actual list
    if isinstance(x, list):
        if len(x) != 1:
            return None
        x = x[0]

    # Case 2: list-like string: "['frame_action']"
    if isinstance(x, str):
        try:
            val = ast.literal_eval(x)
            if isinstance(val, list) and len(val) == 1:
                x = val[0]
        except Exception:
            pass

    if not isinstance(x, str):
        return None

    x = x.strip().lower()
    if x.startswith("frame_"):
        x = x[len("frame_"):]
    return x


primary_frames = (
    pd.concat([src, tgt])
      .map(normalize_frame_name)
      .dropna()
      .unique()
)

primary_set = set(primary_frames)

with open("data/primary_frames.txt", "w", encoding="utf-8") as fw:
    for f in sorted(primary_set):
        fw.write(f + "\n")

print(f"Wrote {len(primary_set)} frames to primary_frames.txt")
```

```python
# === 2) filter the wordlists file by the primary frames ===
import re

INPUT = "data/word_lists.txt"                   # original file (Frame_...: format)
OUTPUT = "data/word_lists_primary_only.txt"     # filtered output

frame_header_re = re.compile(r'^Frame_(.+?):\s*$')
pos_re = re.compile(r'\.[A-Za-z]+$')            # .n .v .a etc.
lead_marker_re = re.compile(r'^[\-\*\•\s]+')    # drop bullets/spaces

with open(INPUT, "r", encoding="utf-8") as fin, \
     open(OUTPUT, "w", encoding="utf-8") as fout:

    current_frame = None
    include_current = False
    words = []
    seen = set()

    for raw in fin:
        line = raw.strip()
        if not line:
            continue

        m = frame_header_re.match(line)
        if m:
            # flush previous frame if included
            if include_current and words:
                fout.write(current_frame + "\n")
                for w in words:
                    fout.write(w + "\n")
                fout.write("\n")

            current_frame = m.group(1).lower()
            include_current = current_frame in primary_set
            words = []
            seen = set()
            continue

        if not include_current:
            continue

        token = lead_marker_re.sub('', line)
        # skip multiword entries
        if " " in token:
            continue
        token = pos_re.sub('', token).strip()
        if not token:
            continue
        if token in seen:
            continue
        seen.add(token)
        words.append(token)

    # final flush
    if include_current and words:
        fout.write(current_frame + "\n")
        for w in words:
            fout.write(w + "\n")
        fout.write("\n")

print("Wrote filtered wordlists to:", OUTPUT)
```

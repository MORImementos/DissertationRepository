Generate word lists from MetaNet for use in queries

```python
import pandas as pd
```

```python
lu_df = pd.read_csv("../metanet/data/trimmed/LexicalUnit.csv")
```

```python
frame_df = pd.read_csv("../metanet/data/trimmed/Frame.csv")
```

```python
frame_and_lu = frame_df[['name', 'hasLexicalUnit']]
```

```python
frame_and_lu = frame_and_lu.dropna()
frame_and_lu
```

```python
import ast
frame_and_lu['hasLexicalUnit'] = frame_and_lu['hasLexicalUnit'].apply(ast.literal_eval)
def get_lu_names(lu_list):
    return lu_df.loc[lu_df['name'].isin(lu_list), 'name'].tolist()

frame_and_lu['lu_names'] = frame_and_lu['hasLexicalUnit'].apply(get_lu_names)
```

```python
frame_and_lu
```

```python
frame_expanded = frame_and_lu.explode('lu_names')
frame_expanded
```

```python
lus = {}
for x, row in frame_expanded.iterrows():
    n = row['name']
    lu = row['lu_names']
    if n not in lus:
        lus[n] = []
    lu_row = lu_df[lu_df['name'] == lu]
    lem = lu_row.hasLemma.values[0]
    lem = lem.replace("['", "").replace("']", "")
    lus[n].append(lem)
```

```python
lus
```

```python
def process_lemmas(lexical_units_dict):
    lud = lexical_units_dict
    p_dict = {}
    for k, v in lud.items():
        if k not in p_dict:
            p_dict[k] = []
        for i in v:
            split_lemma = tuple(i.split('.'))
            p_dict[k].append(split_lemma[0])
    return p_dict



```

```python
lemma_lists = process_lemmas(lus)
lemma_lists
```

```python
with open("data/word_lists.txt", "w", encoding="utf-8") as f:
    for k, v in lus.items():
        f.write(f"{k}:\n")
        for item in v:
            f.write(f"  {item}\n")
```

Word lists have been generated. The next step(s) is to utilize the wordlists + constructions to generate queries, as well as add the wordlists themselves.

```python
import re
from collections import defaultdict

INPUT = "data/word_lists.txt"

frame_header_re = re.compile(r'^Frame_(.+?):\s*$')
pos_re = re.compile(r'\.[A-Za-z]+$')
lead_marker_re = re.compile(r'^[\-\*\•\s]+')

frames = defaultdict(list)
seen_per_frame = defaultdict(set)

current_frame = None

with open(INPUT, "r", encoding="utf-8") as fin:
    for raw in fin:
        line = raw.strip()
        if not line:
            continue

        m = frame_header_re.match(line)
        if m:
            current_frame = m.group(1).lower()
            continue

        if current_frame is None:
            continue

        token = lead_marker_re.sub('', line)
        if " " in token:
            continue

        token = pos_re.sub('', token).strip()
        if not token:
            continue

        if token not in seen_per_frame[current_frame]:
            seen_per_frame[current_frame].add(token)
            frames[current_frame].append(token)

num_frames = len(frames)
print(num_frames)
frames_sorted = sorted(
    frames.items(),
    key=lambda item: len(item[1]),
    reverse=True
)
for frame, words in frames_sorted:
    print(frame, len(words))

OUTPUT = "data/word_lists_cleaned_sorted.txt"

with open(OUTPUT, "w", encoding="utf-8") as fout:
    for frame, words in frames_sorted:
        fout.write(frame + "\n")
        for w in words:
            fout.write(w + "\n")
        fout.write("\n")

```
